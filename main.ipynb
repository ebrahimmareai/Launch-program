# This is the 'main' notebook.
It acts as the true glue layer that connects the 'business_logic' notebook and the 'user_interface' notebook. It imports both, forces them to reload the latest code, creates the logic instance, creates the main window, and passes the logic into the user interface to connect them perfectly.


```python
import tkinter as tk # crea application window
import import_ipynb # import ipynb files as modules



import business_logic 
import user_interface 
 # import business logic and user interface

logic_instance = business_logic.BusinessLogic() # creat business logic
root_window = tk.Tk() # creat the main window 

app = user_interface.UserInterface(root_window, logic_instance) # build UI and link it with business logic



app.build_edit_group_section() # add edit group bush putton
app.edit_group_button.configure(command=app.open_edit_group_window) # link the edite Group button with click function

if hasattr(app, 'create_group_button'): # make sure edite group work
    app.create_group_button.configure(command=app.on_create_group_clicked) # link it with logic function

root_window.mainloop() # open the application
```
