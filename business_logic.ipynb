```python
import webbrowser  # open URLs
import os  # open local file paths
import platform  # check the operating system
import subprocess  # cross-platform command execution
import urllib.parse  # analyze web URLs
from tkinter import filedialog  # open the file explorer window
```

# This notebook will include: 
adding program paths, webpage URLs, PDF files, and DOCX files; creating groups with custom names; adding multiple items to the same group; creating separate groups for different activities; editing or deleting group items; and opening all items in a selected group. Support file types: Only PDF and DOCX formats for both source and destination.

# We are adding the logic for the file explorer. 
This keeps the visual design completely separate from the data and logic.

```python
class BusinessLogic: # define the main class
    def __init__(self): # define new method 
        self.groups = {} # creat an empty dictionary 

    def get_summary_name(self, input_string): # take a path or URL 
        input_string = input_string.strip() # remove spaces 
        
        if input_string.startswith("http://") or input_string.startswith("https://"): # check if user input URL
            parsed_url = urllib.parse.urlparse(input_string) # break down URL
            netloc = parsed_url.netloc # extract main part 
            
            if not netloc: # check if the main part empty
                netloc = "Website" # select fallback name 
            
            clean_name = netloc.replace("www.", "").replace(".com", "").replace(".org", "").replace(".net", "").replace(".", " ").title() # show clear name 
            return clean_name # Returning the cleaned up website name
            
        else: # If not URL it is a local file path 
            file_name_with_ext = os.path.basename(input_string) # extract file name
            clean_name = os.path.splitext(file_name_with_ext)[0] # get just the name
            
            if not clean_name: # check if name empty 
                clean_name = "File" # select fallback name 
            
            return clean_name # run clean file name


    def open_file_explorer(self): # define function tto open explorer window
        file_path = filedialog.askopenfilename(title="Select a File", filetypes=[("All Supported", "*.exe *.pdf *.docx *.lnk"), ("All Files", "*.*")]) # type of files
        return file_path # return file to user interface layer
```

# this code will add logic to open all the programs:
PDFs, DOCXs, and web pages.

```python
def launch_items(self, items_list): # function to open all items 
    for item in items_list: # Looping
        parts = item.split(" -> ", 1) # split path name 
        if len(parts) == 2: # check if split successful 
            path_or_url = parts[1].strip() # extract path or URL
            
            if path_or_url.startswith("http://") or path_or_url.startswith("https://"): # check if item is URL
                webbrowser.open(path_or_url) # open URL in web browser
            else: # if not URL it is local file 
                try: # use try-except to prevent the program for stopping
                    if platform.system() == 'Windows': # check if system open as windows
                        os.startfile(path_or_url) # open local file as system application 
                    elif platform.system() == 'Darwin': # check if operating system macOS
                        subprocess.call(('open', path_or_url)) # open file use open command on macOS
                    else: # Assuming Linux for any other operating system
                        subprocess.call(('xdg-open', path_or_url)) # open file use command on Linux
                except Exception as e: # catch any error
                    print(f"Could not open {path_or_url}: {e}") # print error message 

BusinessLogic.launch_items = launch_items # add launch logic to business logic 
```

# this code replacing the old launch_items function with a safer version:
that checks if the path or URL is valid before trying to open it. This prevents the error where code text gets accidentally opened as a file.

```python
import webbrowser
import os
import platform
import subprocess

def launch_items_safe(self, items_list): # defined launch function
    for item in items_list: # Looping 
        if " -> " not in item: # check if item has arrow separator
            continue # skip item does not have arrow separator
        parts = item.split(" -> ", 1) # splite string get the path
        path_or_url = parts[1].strip() # extract path or URL
        
        if path_or_url.startswith("#") or path_or_url.startswith("def ") or path_or_url.startswith("import "): # check if the path is actually code
            continue # skipe code text if not a real path
        
        if path_or_url.startswith("http://") or path_or_url.startswith("https://"): # check if item was a web URL
            webbrowser.open(path_or_url) # open URL 
        elif os.path.exists(path_or_url): # check if the local file exists
            try: # use try-except to remove crashes
                if platform.system() == 'Windows': # check if the OS window
                    os.startfile(path_or_url) # open the file use windows application
                elif platform.system() == 'Darwin': # check if the OS is macOS
                    subprocess.call(('open', path_or_url)) # open the file on macOS
                else: # Assuming Linux
                    subprocess.call(('xdg-open', path_or_url)) # open the file on Linux
            except Exception as e: # Catching any errors
                print(f"Could not open {path_or_url}: {e}") # print the error
        else: # if file does not exist and not URL
            print(f"File not found: {path_or_url}") # print message 

# Safety check to prevent NameError if run out of order
if 'BusinessLogic' in globals():
    BusinessLogic.launch_items = launch_items_safe # replace the old launch function with the new
    print("SUCCESS: launch_items_safe added to BusinessLogic!")
else:
    print("WAIT: Please run the FIRST cell (class BusinessLogic:) before running this cell!")
```

# We are adding the logic to create a named group:
that stores the items (programs, PDFs, DOCXs, URLs) the user has added. The group is saved in the self.groups dictionary with the group name as the key.


```python
def create_group(self, group_name, items_list): # function to create a new group 
    if not group_name.strip(): # check group name is empty 
        return False # return false if group was not created
    if not items_list: # check if list is empty
        return False # return False if group not created
    self.groups[group_name.strip()] = list(items_list) # save group in dictionary 
    return True # return true if group created 

def get_all_groups_names(self): # define function to return group names
    return list(self.groups.keys()) # return groups dictionary as list

BusinessLogic.create_group = create_group # add create group logic to business logic
BusinessLogic.get_all_groups_names = get_all_groups_names # add get group name logic to business logic 
```

# this code add the logic to get the items in a group:
also we update a group with new items. This allows the user to edit groups by adding or removing items.

```python

def get_group_items(self, group_name): # define function get items from group
    if group_name in self.groups: # exists groups dictionary
        return list(self.groups[group_name]) # return copy group's items as list
    return [] # return empty list 

def update_group_items(self, group_name, new_items): # update a group with new items
    if group_name in self.groups: # check if group exists in groups dictionary
        self.groups[group_name] = list(new_items) # replace the group's items withnew items
        return True # return True if update was successful
    return False # return False if update don't successful

def rename_group(self, old_name, new_name): # function to rename group
    if old_name in self.groups: # check if old group name exists
        if new_name.strip() and new_name not in self.groups: # check if new name valid 
            self.groups[new_name] = self.groups.pop(old_name) # rename group
            return True # return True if rename was successful
    return False # return False if the rename failed

BusinessLogic.get_group_items = get_group_items # add get group items to business logic 
BusinessLogic.update_group_items = update_group_items # add update group items logic business logic 
BusinessLogic.rename_group = rename_group # add rename group logic to business logic 
```

# this code add the logic to delete a group
completely from the groups dictionary.

```python
def delete_group(self, group_name): # define function to delete group
    if group_name in self.groups: # check if group in groups dictionary
        del self.groups[group_name] # remove group from the dictionary
        return True # return True if delete successful
    return False # return False if delete not successful

BusinessLogic.delete_group = delete_group # add delete group to business logic 
```

# this code add the logic to launch all items in a specific group:
This reuses the existing launch_items function to open all programs, PDFs, DOCXs, and URLs.

```python
def launch_group(self, group_name): # define function launch items 
    if group_name in self.groups: # check if group in groups dictionary
        items = self.groups[group_name] # get all items in group
        if items: # check if group has item
            self.launch_items(items) # call launch_items function to open item
            return True # return True if launch successful
        return False # return False if launch not successful
    return False # return False if lanch not successful

BusinessLogic.launch_group = launch_group # add launch group to business logic 
```
